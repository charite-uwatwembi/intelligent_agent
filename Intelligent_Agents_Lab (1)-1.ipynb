# AI for Engineers: Intro to Intelligent Agents

Welcome to your first hands-on AI lab! In this notebook, we are going to build a simple **Intelligent Agent**.

In AI, an **agent** is anything that can perceive its environment through **sensors** and act upon that environment through **actuators**.

To keep things simple, we will build a simulated robotic vacuum cleaner (like a Roomba).
- **The Environment:** A hallway made of a few rooms. Some are clean, some are dirty.
- **The Sensors:** The robot can detect if the room it is currently standing in is dirty or clean.
- **The Actuators:** The robot can suck up dirt, or it can drive forward to the next room.

To run the code blocks below, click the **'Play'** button on the left of the cell, or select the cell and press **Shift + Enter**.

## Part 1: Setting up the Environment
Before we build the robot, we need a world for it to live in. We will use a Python `list` to represent a hallway of 5 rooms.

We will use the number `1` to represent a **Dirty** room, and `0` to represent a **Clean** room. Execute the cell below to create our world.

```python
# 1 = Dirty, 0 = Clean
hallway = [1, 0, 1, 1, 0]

print('The environment has been created!')
print('Current hallway status:', hallway)
```

## Part 2: Building the Agent's "Brain"
Now we need to program the logic of our agent. This is called a **Simple Reflex Agent**. It doesn't have memory; it just looks at what is happening *right now* and reacts to it based on a rule.

Our rules are very simple:
1. IF the room is dirty (`1`), THEN clean it.
2. IF the room is already clean (`0`), THEN move right to the next room.

```python
def robot_brain(room_status):
    """This function takes in what the robot sees, and outputs an action."""
    if room_status == 1:
        return 'CLEAN'
    elif room_status == 0:
        return 'MOVE_RIGHT'

print('Robot brain successfully programmed!')
```

### Question 2:

**Answer:**

An intelligent agent should align its actions with its objective

## Part 3: Running the Simulation
Now we put the robot in the hallway!

We will use a `while` loop. This loop acts as the passage of time. In every "tick" (iteration) of the loop, the robot will:
1. Check the status of its current room.
2. Send that status to its brain to decide on an action.
3. Execute the action (either change the room to a `0`, or move to the next room).

Run the code below and watch the output!

```python
# Reset the hallway just in case
hallway = [1, 0, 1, 1, 0]
agent_position = 0  # The robot starts at the very first room (index 0)
step_counter = 1

print('--- STARTING SIMULATION ---')

while agent_position < len(hallway):
    print(f'\nStep {step_counter}:')

    # 1. Sense the environment
    current_room_status = hallway[agent_position]
    print(f'Robot is in room {agent_position}. The room is {current_room_status}.')

    # 2. Decide on an action
    action = robot_brain(current_room_status)
    print(f'Robot decides to: {action}')

    # 3. Execute the action
    if action == 'CLEAN':
        hallway[agent_position] = 0
        print('Robot vacuumed the dirt!')
    elif action == 'MOVE_RIGHT':
        agent_position = agent_position + 1
        print('Robot moved to the next room.')

    print(f'Current hallway: {hallway}')
    step_counter += 1

print('\n--- SIMULATION FINISHED ---')
print('The whole hallway is clean!')
```

**Answer:**


# Optional Challenge Part 2: The Noisy Sensor (Uncertainty in AI)

So far, our agent has operated in a **Deterministic** environment. When it looked at a room, it knew with 100% certainty if it was dirty or clean.

However, as engineers, we know physical sensors are rarely perfect. Dust on the camera lens, low lighting, or electrical interference can cause faulty readings. When an environment includes randomness or uncertainty, we call it **Stochastic**.

**Your Research Challenge:**
You need to simulate a cheap, faulty dirt sensor that is only **80% accurate**. The other 20% of the time, the sensor errors out and returns `'UNKNOWN'`.

**Your Tasks:**
1. **Do some research:** Look up how to generate random numbers or probabilities in Python. (Hint: Research the `random` module, specifically functions that can help you trigger an event exactly 20% of the time).
2. **Build the faulty sensor:** Write a function called `sense_environment(actual_room_status)`. It should use your researched method to return the true status 80% of the time, and return `'UNKNOWN'` 20% of the time.
3. **Update the Brain:** Modify your `robot_brain` from earlier. How should your agent react if the room status is `'UNKNOWN'`? Should it risk cleaning a clean room? Should it skip it? You decide the logic!
4. **Run the simulation:** Put it all together in a loop and observe how your robot handles a noisy, uncertain world.


```python
import random

def sense_environment(actual_room_status):
    """
    Simulates a faulty sensor.
    80% chance to return actual_room_status.
    20% chance to return 'UNKNOWN'.
    """
    if random.random() < 0.8:
        return actual_room_status
    else:
        return 'UNKNOWN'


def robust_robot_brain(sensor_reading):
    """
    Takes in the sensor reading (1, 0, or 'UNKNOWN') and outputs an action.
    Strategy: Be conservative with unknown readings - clean if uncertain.
    """
    if sensor_reading == 1:
        return 'CLEAN'
    elif sensor_reading == 0:
        return 'MOVE_RIGHT'
    elif sensor_reading == 'UNKNOWN':
        return 'CLEAN'  # Conservative approach: assume dirty when uncertain


# --- TESTING YOUR NOISY SIMULATION ---
hallway = [1, 0, 1, 1, 0]
agent_position = 0
step_counter = 1

print('--- NOISY SENSOR SIMULATION ---')

while agent_position < len(hallway):
    print(f'\nStep {step_counter}:')
    
    actual_room_status = hallway[agent_position]
    sensor_reading = sense_environment(actual_room_status)
    print(f'Robot is in room {agent_position}. Sensor reads: {sensor_reading}')
    
    action = robust_robot_brain(sensor_reading)
    print(f'Robot decides to: {action}')
    
    if action == 'CLEAN':
        hallway[agent_position] = 0
        print('Robot vacuumed!')
    elif action == 'MOVE_RIGHT':
        agent_position = agent_position + 1
        print('Robot moved right.')
    
    print(f'Current hallway: {hallway}')
    step_counter += 1

print('\n--- SIMULATION FINISHED ---')
```
